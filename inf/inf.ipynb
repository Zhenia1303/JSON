```python
import requests
import json
from bs4 import BeautifulSoup
```

```python

url="https://egzamin-informatyk.pl/testy-inf02-ee08-sprzet-systemy-sieci/"
response = requests.get(url)
soupped = BeautifulSoup(response.content,"html.parser")
```

```python
status_code = response.status_code
```

```python
resultJSON = []
count = 1
for category in range(1, 41):
    question = soupped.select("div.trescE")[count -1].text
    answer_a = soupped.select(f"div#odpa{count}")[0].text
    answer_b = soupped.select(f"div#odpb{count}")[0].text
    answer_c = soupped.select(f"div#odpc{count}")[0].text
    answer_d = soupped.select(f"div#odpd{count}")[0].text
    count +=1
    question_scrapped = soupped.find(question)
    answer_a_scrapped = soupped.find(answer_a)
    answer_b_scrapped = soupped.find(answer_b)
    answer_c_scrapped = soupped.find(answer_c)
    answer_d_scrapped = soupped.find(answer_d)
    
    resultJSON.append(
        {
            "question": question,
            "answerA": answer_a,
            "answerB": answer_b,
            "answerC": answer_c,
            "answerD": answer_d
        }
    )
```

```python
json = json.dumps(resultJSON, indent=2, ensure_ascii=False)
with open("question.json", "w", encoding="utf-8") as f:
    f.write(json)
```
